```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import h2o
from h2o.estimators import H2OExtendedIsolationForestEstimator, H2ORandomForestEstimator
import mlflow.h2o
import mlflow

import config
```

```python
# define Mlflow env. settings
mlflow.set_tracking_uri(uri=config.mlflow_uri)
%env MLFLOW_TRACKING_URI= config.mlflow_db
%env mlflow_tracking_username = config.mlflow_un
%env mlflow_tracking_password = config.mlflow_pw
mlflow.enable_system_metrics_logging()
mlflow.autolog()

# Initialize H2O server
h2o.init(ip="127.0.0.1", port="8080", nthreads=-1, min_mem_size= config.min_memory, max_mem_size= config.max_memory, enable_assertions=False)

# Clear h2o memory
h2o.remove_all()
```

```python
#mlflow.autolog(log_models=True, log_datasets=True)
mlflow.set_experiment("Extended Isolation Forest Experiment")
customerList = ["X5","X7","X8"]#,"X3","X4","X1","X2","X6"] #
with mlflow.start_run(run_name=f"EIF_Base_ColdStart", nested=True) as grandparent_run:
    for customer in customerList:
        filepath = Path(f'C:/MLAR_DATA/T0001/FS_base_data/FS_base_data_{customer}.parquet')
        if filepath.is_file()==True:
            df = pd.read_parquet(filepath)
            df.dropna(how='all', axis=1, inplace=True)
            h2oFrame = h2o.H2OFrame(python_obj=df)
            #h2oFrame["T0083_ANOMALY"] = h2oFrame["T0083_ANOMALY"].asfactor()
            
            features = h2oFrame.drop(['T0083','T0083_ANOMALY','T0083_SEVERITY']).columns
            train, test = h2oFrame.split_frame(ratios =[0.7], seed=1234)
            
            # Define model parameters
            numTrees = 100

            # Initialize the model
            eif_model = H2OExtendedIsolationForestEstimator(seed=12345)
            
            # Start an MLflow run
            with mlflow.start_run(run_name=f"Predicted_Dataset_{customer}", nested=True) as parent_run:
                # Log model parameters and train
                #mlflow.log_param("ntrees", numTrees)
                #mlflow.log_param("sample_size", sampleSize)
                eif_model.train(x=features,training_frame=train) 
                
                # Predict anomalies and generate histogram
                predictedH2oFrame = eif_model.predict(train)
                predictedDF = predictedH2oFrame.as_data_frame(use_pandas=True, use_multi_thread=True)
                figure, axis = plt.subplots()
                axis.hist(predictedDF["mean_length"])
                mlflow.log_figure(figure, f"{customer}_base_mean_length.png")               
                
                # Combine the scores with the original data
                anomaliesTrain = train[predictedH2oFrame["mean_length"] < 10.0]
                anomaliesTest = test[eif_model.predict(test)["mean_length"] < 10.0]
                finalTrainFrame = eif_model.predict(anomaliesTrain).cbind(anomaliesTrain)
                finalTestFrame = eif_model.predict(anomaliesTest).cbind(anomaliesTest)
                #train, test = predictedH2oFrame.split_frame(ratios =[0.7], seed=1234)

                #Export Dataset
                mlflow.log_table(predictedDF,"predictedDF.json")
                
                with mlflow.start_run(run_name=f"Explain_model", nested=True) as child_run:
                    #Train explaination model
                    DRF_model = H2ORandomForestEstimator(ntrees=numTrees)
                    DRF_model.train(x=features, y="anomaly_score", training_frame=finalTrainFrame) 

                    # DRF_model.explain(finalH2oFrame) equivalent minus pdp, ice plots
                    ra_plot = DRF_model.residual_analysis_plot(finalTestFrame).figure()
                    mlflow.log_figure(ra_plot, f"{customer}_base_residual_analysis_plot.png")
                    vi_plot = plt.figure(figsize=(9, 3))
                    DRF_model.varimp_plot()
                    mlflow.log_figure(vi_plot, f"{customer}_base_variable_importance_plot.png")
                    ss_plot = DRF_model.shap_summary_plot(finalTestFrame).figure()
                    mlflow.log_figure(ss_plot, f"{customer}_base_shap_summary_plot.png") 
                    lc_plot = DRF_model.learning_curve_plot().figure()
                    mlflow.log_figure(lc_plot, f"{customer}_base_learning_curve_plot.png")     

                    #predictContribution = DRF_model.predict_contributions(predictedH2oFrame)
        else:
            print("Error - Feature store file not found. Check directory.")

```
